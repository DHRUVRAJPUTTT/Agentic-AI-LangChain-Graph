### GROQ MODEL INTEGRATION

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Who is the prime minister of India")
response.content
```

### STREAMING !


 model.invoke (Batch method) :

It first generates the entire response , when 100% finished it hands the complete final string

model.stream :-)

Most models can stream their output while it is being generated . By displaying output progressively , streaming significantly improves user experience, particularly for long responses.
Calling stream() returns an iterator that yields output chunks as they are produced.


```python
for chunk in model.stream("Write me a 200 words paragraph on AgenticAI and it's demand."):
    print(chunk.content, end="", flush=True)
```

### BATCH

Batching is a collection of independent requests to a model , can significantly improve performance and reduce costs , as the processing can be done in parallel.

```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
model = ChatGroq(model="llama3-8b-8192")
responses = model.batch([
    "Who was Bhagat Singh ?"
    "When is childrens day celebrated ?"
    "What skills are requried to learn AGENTIC AI ?"
])
for response in responses:
    print(response)
```
